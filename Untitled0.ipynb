```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

def generate_sar_target(target_type, img_size=128):
    """Generates a realistic X-band SAR radar scan with metallic reflections, radar shadow, and speckle noise."""
    # 1. Background terrain clutter (low radar backscatter)
    img = np.random.normal(loc=45, scale=10, size=(img_size, img_size))

    # Randomize target angle (azimuth) and slight position shift
    angle = np.random.uniform(0, 360)
    cx, cy = img_size // 2 + np.random.randint(-8, 8), img_size // 2 + np.random.randint(-8, 8)

    # Create target mask
    target_layer = np.zeros((img_size, img_size), dtype=np.float32)
    shadow_layer = np.zeros((img_size, img_size), dtype=np.float32)

    if target_type == 0:   # Main Battle Tank (T-72 style: wide hull + long turret barrel)
        cv2.rectangle(target_layer, (cx-18, cy-12), (cx+18, cy+12), 180, -1)
        cv2.line(target_layer, (cx, cy), (cx+32, cy), 220, 4) # Gun barrel
        cv2.rectangle(shadow_layer, (cx-18, cy-24), (cx+18, cy-12), -35, -1) # Radar shadow

    elif target_type == 1: # Armored Personnel Carrier (BTR style: elongated 8-wheel chassis)
        cv2.rectangle(target_layer, (cx-24, cy-10), (cx+24, cy+10), 165, -1)
        cv2.circle(target_layer, (cx-5, cy), 6, 210, -1) # Small turret
        cv2.rectangle(shadow_layer, (cx-24, cy-20), (cx+24, cy-10), -35, -1)

    elif target_type == 2: # Mobile Radar / Missile Launcher (Large rectangular array)
        cv2.rectangle(target_layer, (cx-26, cy-14), (cx+26, cy+14), 190, -1)
        cv2.line(target_layer, (cx-20, cy-8), (cx+20, cy+8), 240, 3) # Strong metallic corner reflector
        cv2.rectangle(shadow_layer, (cx-26, cy-28), (cx+26, cy-14), -40, -1)

    elif target_type == 3: # Military Supply Truck (Cab + cargo bed separation)
        cv2.rectangle(target_layer, (cx-22, cy-9), (cx+4, cy+9), 150, -1)  # Cargo bed
        cv2.rectangle(target_layer, (cx+8, cy-9), (cx+20, cy+9), 185, -1) # Metallic cab
        cv2.rectangle(shadow_layer, (cx-22, cy-18), (cx+20, cy-9), -30, -1)

    # Rotate target & shadow to simulate satellite pass angle
    M = cv2.getRotationMatrix2D((cx, cy), angle, 1.0)
    target_layer = cv2.warpAffine(target_layer, M, (img_size, img_size))
    shadow_layer = cv2.warpAffine(shadow_layer, M, (img_size, img_size))

    # Combine terrain + shadow + bright metallic target
    combined = img + shadow_layer + target_layer
    combined = np.clip(combined, 5, 255)

    # 2. Apply authentic multiplicative SAR Speckle Noise (Gamma distribution)
    speckle = np.random.gamma(shape=2.5, scale=0.4, size=(img_size, img_size))
    sar_scan = np.clip(combined * speckle, 0, 255).astype(np.uint8)

    return sar_scan

# Generate a full dataset of 2,000 SAR images (500 per military class)
class_names = ['Main Battle Tank', 'Armored Carrier', 'Missile Launcher', 'Supply Truck']
X_data, y_data = [], []

for class_id in range(4):
    for _ in range(500):
        X_data.append(generate_sar_target(class_id))
        y_data.append(class_id)

X_data = np.array(X_data)
y_data = np.array(y_data)

print(f"Success! Generated {len(X_data)} X-Band SAR Radar images.")
print(f"Dataset shape: {X_data.shape}, Classes: {class_names}")
```

```python
plt.figure(figsize=(14, 4))
for i in range(4):
    # Pick one sample from each military class
    idx = i * 500
    plt.subplot(1, 4, i+1)
    plt.imshow(X_data[idx], cmap='gray')
    plt.title(f"Target: {class_names[y_data[idx]]}")
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
def lee_filter(img, win_size=5):
    """Applies a statistical Lee filter to remove SAR speckle noise while preserving target edges."""
    img = img.astype(np.float32)
    img_mean = cv2.blur(img, (win_size, win_size))
    img_sqr_mean = cv2.blur(img**2, (win_size, win_size))
    img_variance = np.maximum(0, img_sqr_mean - img_mean**2)

    overall_variance = np.var(img)
    img_weights = img_variance / (img_variance + overall_variance + 1e-5)
    img_filtered = img_mean + img_weights * (img - img_mean)
    return np.clip(img_filtered, 0, 255).astype(np.uint8)

# 1. Clean all 2,000 radar images in our dataset
print("Applying Lee Despeckling Filter to all 2,000 SAR scans...")
X_clean = np.array([lee_filter(img) for img in X_data])
print("Despeckling complete!")

# 2. Show a side-by-side comparison (Raw Noisy SAR vs. Cleaned SAR) for your thesis report
plt.figure(figsize=(10, 6))
for i in range(2):
    idx = i * 500
    # Top row: Raw Noisy SAR
    plt.subplot(2, 2, i*2 + 1)
    plt.imshow(X_data[idx], cmap='gray')
    plt.title(f"Raw Noisy SAR: {class_names[y_data[idx]]}")
    plt.axis('off')

    # Bottom row: Lee Filtered SAR
    plt.subplot(2, 2, i*2 + 2)
    plt.imshow(X_clean[idx], cmap='gray')
    plt.title(f"Lee Filtered: {class_names[y_data[idx]]}")
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
from sklearn.model_selection import train_test_split

# 1. Use the T4 GPU if available
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Training AI on hardware: {device}")

# 2. Prepare data for PyTorch (Normalize pixels to 0-1 and add channel dimension)
X_tensor = torch.tensor(X_clean, dtype=torch.float32).unsqueeze(1) / 255.0
y_tensor = torch.tensor(y_data, dtype=torch.long)

# Split into 1,600 training scans and 400 unseen test scans
X_train, X_test, y_train, y_test = train_test_split(
    X_tensor, y_tensor, test_size=0.2, random_state=42, stratify=y_tensor
)

train_loader = DataLoader(TensorDataset(X_train, y_train), batch_size=32, shuffle=True)
test_loader = DataLoader(TensorDataset(X_test, y_test), batch_size=32, shuffle=False)

# 3. Define the Custom SAR Convolutional Neural Network
class SARTargetNet(nn.Module):
    def __init__(self, num_classes=4):
        super(SARTargetNet, self).__init__()
        self.features = nn.Sequential(
            nn.Conv2d(1, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),

            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),

            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),
            nn.AdaptiveAvgPool2d((4, 4))
        )
        self.classifier = nn.Sequential(
            nn.Linear(128 * 4 * 4, 256),
            nn.ReLU(),
            nn.Dropout(0.4),
            nn.Linear(256, num_classes)
        )

    def forward(self, x):
        x = self.features(x)
        x = torch.flatten(x, 1)
        return self.classifier(x)

model = SARTargetNet(num_classes=4).to(device)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

# 4. Train the AI for 15 Epochs
train_losses, test_accuracies = [], []
print("\nStarting AI Training...")

for epoch in range(15):
    model.train()
    running_loss = 0.0
    for inputs, labels in train_loader:
        inputs, labels = inputs.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item()

    # Evaluate on unseen test radar scans
    model.eval()
    correct, total = 0, 0
    with torch.no_grad():
        for inputs, labels in test_loader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            _, predicted = torch.max(outputs.data, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()

    acc = 100 * correct / total
    train_losses.append(running_loss / len(train_loader))
    test_accuracies.append(acc)
    print(f"Epoch [{epoch+1:2d}/15] | Training Loss: {train_losses[-1]:.4f} | Test Accuracy: {acc:.2f}%")

# 5. Save the trained AI model weights
torch.save(model.state_dict(), "sar_classifier.pth")
print("\nTraining Complete! Model brain saved as 'sar_classifier.pth'")
```

```python
import seaborn as sns
from sklearn.metrics import confusion_matrix, classification_report
import torch.nn.functional as F

# 1. Gather all predictions on the 400 unseen test radar scans
model.eval()
all_preds, all_labels = [], []

with torch.no_grad():
    for inputs, labels in test_loader:
        inputs = inputs.to(device)
        outputs = model(inputs)
        _, predicted = torch.max(outputs, 1)
        all_preds.extend(predicted.cpu().numpy())
        all_labels.extend(labels.numpy())

# Print detailed Precision, Recall, and F1-Score report for your thesis table
print("=== DETAILED MILITARY TARGET CLASSIFICATION REPORT ===")
print(classification_report(all_labels, all_preds, target_names=class_names))

# 2. Plot Training Curve & Confusion Matrix Side-by-Side
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Left Plot: Accuracy Progression
axes[0].plot(range(1, 16), test_accuracies, marker='o', color='navy', linewidth=2)
axes[0].set_title("Model Validation Accuracy Over 15 Epochs", fontsize=12, fontweight='bold')
axes[0].set_xlabel("Epoch")
axes[0].set_ylabel("Accuracy (%)")
axes[0].grid(True, linestyle='--', alpha=0.6)

# Right Plot: Confusion Matrix
cm = confusion_matrix(all_labels, all_preds)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names, ax=axes[1])
axes[1].set_title("Confusion Matrix (Unseen Test Scans)", fontsize=12, fontweight='bold')
axes[1].set_xlabel("AI Predicted Target")
axes[1].set_ylabel("Actual Military Target")

plt.tight_layout()
plt.show()

# 3. Explainable AI (Grad-CAM): Visualize WHAT the AI focuses on inside the radar scan
def generate_gradcam(model, img_tensor, target_class):
    model.eval()
    img_tensor = img_tensor.unsqueeze(0).to(device)
    img_tensor.requires_grad = True

    # Hook into the final convolutional layer
    activations, gradients = [], []
    def forward_hook(module, inp, out): activations.append(out)
    def backward_hook(module, grad_in, grad_out): gradients.append(grad_out[0])

    h1 = model.features[8].register_forward_hook(forward_hook)
    h2 = model.features[8].register_full_backward_hook(backward_hook)

    output = model(img_tensor)
    model.zero_grad()
    output[0, target_class].backward()

    h1.remove()
    h2.remove()

    # Compute heatmap weights
    grads = gradients[0].cpu().data.numpy()[0]
    acts = activations[0].cpu().data.numpy()[0]
    weights = np.mean(grads, axis=(1, 2))
    cam = np.zeros(acts.shape[1:], dtype=np.float32)
    for i, w in enumerate(weights):
        cam += w * acts[i]

    cam = np.maximum(cam, 0)
    cam = cv2.resize(cam, (128, 128))
    cam = cam / (np.max(cam) + 1e-8)
    return cam

# Plot Grad-CAM XAI Heatmaps for 4 test vehicles
plt.figure(figsize=(14, 4))
for i in range(4):
    # Find one test image per class
    idx = (y_test == i).nonzero(as_tuple=True)[0][0].item()
    sample_img = X_test[idx]
    heatmap = generate_gradcam(model, sample_img, i)

    plt.subplot(1, 4, i+1)
    plt.imshow(sample_img.squeeze(), cmap='gray')
    plt.imshow(heatmap, cmap='jet', alpha=0.45) # Overlay red/yellow AI attention map
    plt.title(f"XAI Focus: {class_names[i]}")
    plt.axis('off')

plt.suptitle("Explainable AI (Grad-CAM): Red/Yellow Areas Show Where the AI Detected Target Signatures", fontsize=12, fontweight='bold', y=1.05)
plt.tight_layout()
plt.show()
```
